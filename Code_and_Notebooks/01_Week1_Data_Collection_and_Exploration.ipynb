```python
# Import required libraries

import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries imported successfully.")
```

```python
# Define project directories

from pathlib import Path

# Define project root directory
PROJECT_ROOT = Path(r"C:\Users\Admin\Capstone_Project")

# Define data directories
RAW_DATA = PROJECT_ROOT / "Data" / "Raw_Data"
CROSSWALKS = PROJECT_ROOT / "Data" / "Crosswalks"

# Define output directories
OUTPUTS = PROJECT_ROOT / "Outputs"
TABLES = OUTPUTS / "Tables"
FIGURES = OUTPUTS / "Figures"

# Create output folders if they do not exist
for folder in [OUTPUTS, TABLES, FIGURES]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project Root:", PROJECT_ROOT)
print("Raw Data Folder:", RAW_DATA)
print("Crosswalk Folder:", CROSSWALKS)
print("Output folders created successfully.")
```

```python
# Verify data files and folders

print("Raw Data Contents:")
for item in RAW_DATA.iterdir():
    print("-", item.name)

print("\nCrosswalk Contents:")
for item in CROSSWALKS.iterdir():
    print("-", item.name)
```

```python
# Define Job Bank dataset path

JOB_BANK_FILE = (
    RAW_DATA /
    "job_bank-open-data-all-job-postings-june-2026.csv"
)

# Check whether the file exists
if JOB_BANK_FILE.exists():
    print("Job Bank file found.")
    print(JOB_BANK_FILE)
else:
    print("ERROR: Job Bank file not found.")
```

```python
# Load Job Bank dataset

job_bank = pd.read_csv(
    JOB_BANK_FILE,
    low_memory=False
)

print("Job Bank dataset loaded successfully.")
print("Dataset shape:", job_bank.shape)
```

```python
# Check Job Bank dataset size

print("Number of rows:", job_bank.shape[0])
print("Number of columns:", job_bank.shape[1])

print(f"\nDataset contains {job_bank.shape[0]:,} rows and {job_bank.shape[1]} columns.")
```

```python
# Display first and last few records of the dataset

print("First 5 records:")
display(job_bank.head())

print("\nLast 5 records:")
display(job_bank.tail())
```

```python
# Display dataset column names

print(f"Total number of columns: {len(job_bank.columns)}\n")

for i, column in enumerate(job_bank.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Create a summary table of dataset column data types

data_types = pd.DataFrame({
    "Column": job_bank.columns,
    "Data_Type": job_bank.dtypes.astype(str).values
})

data_types
```

```python
data_types.to_csv(
    TABLES / "job_bank_data_types.csv",
    index=False
)
```

```python
# Save data type summary table

data_types.to_csv(
    TABLES / "job_bank_data_types.csv",
    index=False
)

print("Data type summary saved successfully.")
```

```python
# Display dataset information including missing values

job_bank.info()
```

```python
# Check for duplicate records

duplicate_count = job_bank.duplicated().sum()

duplicate_percentage = (
    duplicate_count / len(job_bank)
) * 100

print("Total duplicate rows:", duplicate_count)
print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
# Analyze missing values in the dataset

missing_table = pd.DataFrame({
    "Column": job_bank.columns,
    "Missing_Count": job_bank.isna().sum().values,
    "Missing_Percentage": (
        job_bank.isna().mean() * 100
    ).round(2).values
})

# Sort columns by highest missing percentage
missing_table = missing_table.sort_values(
    by="Missing_Percentage",
    ascending=False
)

# Display columns with highest missing values
missing_table.head(20)
```

```python
# Save missing value summary table

missing_table.to_csv(
    TABLES / "job_bank_missing_value_table.csv",
    index=False
)

print("Missing value summary saved successfully.")
```

```python
# Analyze number of unique values in each column

unique_values = pd.DataFrame({
    "Column": job_bank.columns,
    "Unique_Values": job_bank.nunique().values
})

# Sort columns by number of unique values
unique_values = unique_values.sort_values(
    by="Unique_Values",
    ascending=False
)

unique_values
```

```python
# Save unique value summary table

unique_values.to_csv(
    TABLES / "job_bank_unique_values.csv",
    index=False
)

print("Unique values summary saved successfully.")
```

```python
# Create data dictionary for Job Bank dataset

data_dictionary = pd.DataFrame({
    "Original_Variable_Name": job_bank.columns,
    "Data_Type": job_bank.dtypes.astype(str).values,
    "Unique_Values": job_bank.nunique().values,
    "Missing_Count": job_bank.isna().sum().values,
    "Missing_Percentage": (
        job_bank.isna().mean() * 100
    ).round(2).values,

    "Cleaned_Variable_Name": [
        column.strip()
        .lower()
        .replace(" ", "_")
        for column in job_bank.columns
    ],

    "Definition": "",
    "Unit": "",
    "Missing_Value_Treatment": "",
    "Transformation": "",
    "Intended_Analytical_Use": ""
})

data_dictionary.head()
```

```python
# Save data dictionary as Excel file

data_dictionary.to_excel(
    TABLES / "Job_Bank_Data_Dictionary.xlsx",
    index=False
)

print("Data dictionary saved successfully.")
```

```python
# Create overall dataset summary

dataset_summary = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Duplicate Rows",
        "Duplicate Percentage",
        "Total Missing Cells",
        "Total Missing Percentage"
    ],

    "Value": [
        f"{len(job_bank):,}",
        f"{len(job_bank.columns):,}",
        f"{job_bank.duplicated().sum():,}",
        f"{(job_bank.duplicated().sum() / len(job_bank) * 100):.2f}%",
        f"{job_bank.isna().sum().sum():,}",
        f"{(job_bank.isna().sum().sum() / (len(job_bank) * len(job_bank.columns)) * 100):.2f}%"
    ]
})

dataset_summary
```

```python
# Save dataset summary table

dataset_summary.to_csv(
    TABLES / "job_bank_dataset_summary.csv",
    index=False
)

print("Dataset summary saved successfully.")
```

```python
# Create project run information for reproducibility

from datetime import datetime

run_information = pd.DataFrame({
    "Project": [
        "AI-Powered Career and Skill-Gap Recommendation System"
    ],
    "Dataset": [
        "Government of Canada Job Bank"
    ],
    "Dataset_File": [
        JOB_BANK_FILE.name
    ],
    "Analysis_Run_Date": [
        datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    ],
    "Rows": [
        job_bank.shape[0]
    ],
    "Columns": [
        job_bank.shape[1]
    ],
    "Duplicate_Rows": [
        job_bank.duplicated().sum()
    ],
    "Missing_Cells": [
        job_bank.isna().sum().sum()
    ]
})

run_information
```

```python
# Save run information

run_information.to_excel(
    TABLES / "Job_Bank_Run_Information.xlsx",
    index=False
)

print("Run information saved successfully.")
```
